```python
# We will do a statisical test: Whiskey vs non-whiskey
# Ho(Null Hypothesis): Average price per bottle for whiskey = average price for
# non whiskey.
# Ha(Alternative Hypothesis): Average price per bottle per whiskey not equal
# average price for non whiskey
# We can do a two sample t-test (Welch's t-test)

from scipy import stats

df_raw = pd.read_csv("BigQuery_IowaLiquorSalesData.csv")
df_raw["date"] = pd.to_datetime(df_raw["date"])

cols = [
    "date", "store_name", "city", "county",
    "category_name", "vendor_name", "item_description",
    "bottles_sold", "sale_dollars"
]

df_sel = df_raw[cols].copy()
df = df_sel[df_sel["sale_dollars"] > 0].copy().reset_index(drop=True)

# Price per bottle at row level
df["price_per_bottle"] = df["sale_dollars"] / df["bottles_sold"]

# Flag whiskey rows
whiskey_keywords = ["WHISKEY", "WHISKIES", "BOURBON", "SCOTCH"]
mask = df["category_name"].str.contains("|".join(whiskey_keywords), case=True, na=False)
df["is_whiskey"] = mask

# Separate groups
whiskey_prices = df[df["is_whiskey"]]["price_per_bottle"]
non_whiskey_prices = df[~df["is_whiskey"]]["price_per_bottle"]

# Summary stats
print("Whiskey:")
print("  N:", len(whiskey_prices))
print("  Mean price per bottle:", whiskey_prices.mean())
print("  Std dev:", whiskey_prices.std())

print("\nNon-whiskey:")
print("  N:", len(non_whiskey_prices))
print("  Mean price per bottle:", non_whiskey_prices.mean())
print("  Std dev:", non_whiskey_prices.std())

# Two-sample t-test (Welch)
t_stat, p_val = stats.ttest_ind(whiskey_prices, non_whiskey_prices, equal_var=False)

print("\nTwo-sample t-test (Welch):")
print("  t-statistic:", t_stat)
print("  p-value:", p_val)

alpha = 0.05
if p_val < alpha:
    print("\nResult: Reject H0 – there is a statistically significant difference in average price per bottle.")
else:
    print("\nResult: Fail to reject H0 – no statistically significant difference in average price per bottle.")

# For business people whiskey is not only the largest revenue category but it
# also has a significant higher average price per bottle. So for store owners,
# this strategy of prioritizing whiskey products in inventory, self placement,
# and promotions would maximize profit and revenue.
```

```python
cols = [
    "date", "store_name", "city", "county",
    "category_name", "vendor_name", "item_description",
    "bottles_sold", "sale_dollars"
]

df_sel = df_raw[cols].copy()
df = df_sel[df_sel["sale_dollars"] > 0].copy().reset_index(drop=True)

# Price per bottle at row level
df["price_per_bottle"] = df["sale_dollars"] / df["bottles_sold"]

# Define origin: American vs Imported vs Other
def assign_origin(cat):
    if pd.isna(cat):
        return "Other"
    cat_upper = cat.upper()
    # Imported indicators
    imported_keywords = ["IMPORTED", "SCOTCH", "IRISH", "TEQUILA", "COGNAC", "BRANDY"]
    for k in imported_keywords:
        if k in cat_upper:
            return "Imported"
    # American indicators
    american_keywords = ["AMERICAN", "BOURBON", "TENNESSEE", "RYE"]
    for k in american_keywords:
        if k in cat_upper:
            return "American"
    return "Other"

df["origin"] = df["category_name"].apply(assign_origin)

# Aggregate by origin
origin_agg = (
    df.groupby("origin")[["sale_dollars", "bottles_sold"]]
    .agg(
        total_revenue=("sale_dollars", "sum"),
        total_bottles=("bottles_sold", "sum")
    )
    .reset_index()
)

origin_agg["avg_price_per_bottle"] = (
    origin_agg["total_revenue"] / origin_agg["total_bottles"]
)

print("Revenue and price by origin:")
display(origin_agg)

# Two-sample t-test: American vs Imported price per bottle
american_prices = df[df["origin"] == "American"]["price_per_bottle"]
imported_prices = df[df["origin"] == "Imported"]["price_per_bottle"]

t_stat, p_val = stats.ttest_ind(american_prices, imported_prices, equal_var=False)

print("\nAmerican vs Imported price per bottle:")
print("American mean price (row-level):", american_prices.mean())
print("Imported mean price (row-level):", imported_prices.mean())
print("t-statistic:", t_stat)
print("p-value:", p_val)

alpha = 0.05
if p_val < alpha:
    print("\nResult: Reject H0 – there is a statistically significant difference in average price per bottle.")
else:
    print("\nResult: Fail to reject H0 – no statistically significant difference in average price per bottle.")
```
